```python
df = pd.read_csv(
    "../data/raw/Superstore.csv",
    sep=";",
    encoding="latin1",
    engine="python"
)
df.head()
```

```python
df.shape
df.info()
df.describe()
df.columns
```

```python
df.isnull().sum()
df.duplicated().sum()
df = df.drop_duplicates()

df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    dayfirst=True
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],
    dayfirst=True
)

df.info()
```

```python
df["Month"] = df["Order Date"].dt.month_name()
df["Year"] = df["Order Date"].dt.year
df["Month Number"] = df["Order Date"].dt.month
df["Quarter"] = df["Order Date"].dt.quarter
```

```python
# Convertir columnas numéricas
for col in ["Sales", "Profit", "Discount"]:
    df[col] = (
        df[col]
        .str.replace(",", ".", regex=False)
        .astype(float)
    )
```

```python
df[["Sales", "Profit", "Discount"]].dtypes
```

```python
df[["Sales", "Profit", "Discount"]].head(10)
```

```python
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Order ID"].nunique()
total_customers = df["Customer ID"].nunique()
total_products = df["Product ID"].nunique()

average_order = total_sales / total_orders

print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Orders: {total_orders}")
print(f"Customers: {total_customers}")
print(f"Products: {total_products}")
print(f"Average Order Value: ${average_order:,.2f}")
```

```python
df.groupby("Category")["Sales"].sum().sort_values()
df.groupby("Category")["Profit"].sum().sort_values()

df.groupby("Product Name")["Sales"]\
    .sum()\
    .sort_values(ascending=False)\
    .head(10)

df.groupby("Customer Name")["Sales"]\
    .sum()\
    .sort_values(ascending=False)\
    .head(10)

df.groupby("Region")["Sales"].sum()

df.groupby("State")["Sales"]\
    .sum()\
    .sort_values(ascending=False)\
    .head(10)

df.groupby("State")["Profit"]\
    .sum()\
    .sort_values(ascending=False)\
    .head(10)
```

```python
df.to_csv(
    "../data/processed/superstore_clean.csv",
    index=False
)
```
